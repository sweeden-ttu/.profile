## Harness helpers

```python
import asyncio, concurrent.futures, yaml
import pandas as pd
import litellm
from adk_submission import VllmConfig, VllmServer, discover_adapters
from google.adk.agents.context_cache_config import ContextCacheConfig
from google.adk.apps._configs import EventsCompactionConfig
from swegemma.config import EvalConfig, build_submission_limits, ALLOWED_ADAPTER_EXTENSIONS
from swegemma.evaluate import Evaluator
from swegemma.models import Task

litellm.drop_params = True
MODELS = {"31B": "gemma-4-31b-it-qat-w4a16-ct", "12B": "gemma-4-12b-it-qat-w4a16-ct", "E4B": "gemma-4-e4b-it-qat-w4a16-ct"}
VLLM_EXTRA = {"gemma-4-12b-it-qat-w4a16-ct": ("--hf-overrides", '{"architectures": ["Gemma4ForCausalLM"], "model_type": "gemma4"}')}
SPECS = WORKSPACE / "challenge" / "specs"
AGENTS = WORKSPACE / "challenge" / "agents"


def run_sync(fn):
    try:
        loop = asyncio.get_running_loop()
    except RuntimeError:
        loop = None
    if loop is not None and loop.is_running():
        with concurrent.futures.ThreadPoolExecutor(max_workers=1) as pool:
            return pool.submit(lambda: asyncio.run(fn())).result()
    return asyncio.run(fn())


def agent_dir(role, model):
    d = OUT / "agents" / f"{role}_{model}"
    if d.exists():
        shutil.rmtree(d)
    shutil.copytree(AGENTS / role, d)
    for y in d.rglob("*.yaml"):
        y.write_text(re.sub(r"^(\s*model:\s*).*$", rf"\g<1>{model}", y.read_text(), flags=re.M))
    return d


def make_task(task_id, spec, paths, extra_files=None):
    repo = OUT / "repos" / task_id
    if repo.exists():
        shutil.rmtree(repo)
    repo.mkdir(parents=True)
    (repo / "TASK.md").write_text((SPECS / spec).read_text())
    (repo / "CHALLENGE.md").write_text((WORKSPACE / "challenge" / "CHALLENGE.md").read_text())
    (repo / "ARENA_PATHS.json").write_text(json.dumps({k: str(v) for k, v in paths.items()}, indent=1))
    for name, src in (extra_files or {}).items():
        shutil.copy(src, repo / name)
    git = lambda *a: subprocess.run(["git", *a], cwd=repo, check=True, capture_output=True, text=True).stdout.strip()
    git("init", "-q", "-b", "main")
    git("-c", "user.email=arena@local", "-c", "user.name=arena", "add", "-A")
    git("-c", "user.email=arena@local", "-c", "user.name=arena", "commit", "-q", "-m", "task")
    commit = git("rev-parse", "HEAD")
    snaps = OUT / "snapshots"
    snaps.mkdir(exist_ok=True)
    subprocess.run(["tar", "czf", str(snaps / f"{task_id}.tgz"), "-C", str(repo), "."], check=True)
    statement = ((SPECS / spec).read_text() + "\n\nAll input and output paths for this task are in /workspace/ARENA_PATHS.json. "
                 "The competition rules are in /workspace/CHALLENGE.md.")
    return Task(instance_id=task_id, repo="scottweeden/knee-usd-arena", base_commit=commit, problem_statement=statement)


def run_tasks(role, label, tasks, after_each=None):
    model = MODELS[label]
    adir = agent_dir(role, model)
    mpath = first_dir(f"/kaggle/input/models/google/gemma-4/other/{model}/2", f"/kaggle/input/gemma-4/other/{model}/2")
    assert mpath, f"model {model} is not attached"
    adapters = discover_adapters(str(adir), adapter_extensions=ALLOWED_ADAPTER_EXTENSIONS)
    bf16 = torch.cuda.get_device_capability(0)[0] >= 8
    cfg = VllmConfig(model=str(mpath), port=8000, host="127.0.0.1", tool_call_parser="gemma4", reasoning_parser="gemma4", max_model_len=32768 if bf16 else 16384,
                     dtype="bfloat16" if bf16 else "float16", gpu_memory_utilization=0.90, enable_auto_tool_choice=True, enable_lora=False, max_loras=1, max_lora_rank=128,
                     scheduler_reserve_full_isl=False, tensor_parallel_size=4 if torch.cuda.device_count() >= 4 else max(1, torch.cuda.device_count()),
                     startup_timeout=60 * 20, extra_args=VLLM_EXTRA.get(model, ()))
    server = VllmServer(cfg, adapter_manifest=adapters)
    t0 = time.time()
    server.start()
    print(f"[{label}] vLLM up in {time.time() - t0:.0f}s")
    results = []
    try:
        models = server.create_model_registry(aliases=[model], model_prefix="openai/", api_key="EMPTY")
        ev = dict((yaml.safe_load((adir / "eval_config.yaml").read_text()) or {}).get("evaluation", {}), **BUDGET_OVERRIDE)
        limits, gen = build_submission_limits()
        for task in tasks:
            res_dir = OUT / "results" / label / task.instance_id
            ecfg = EvalConfig(tasks_path=OUT / "tasks.jsonl", snapshots_dir=OUT / "snapshots", results_dir=res_dir, submission_dir=adir, models=models,
                              sandbox="subprocess", timeout_seconds=int(ev.get("timeout_seconds", 900)), verification_timeout_seconds=60,
                              max_time_minutes=float(ev.get("max_time_minutes", 55)), max_tool_calls=int(ev.get("max_tool_calls", 120)),
                              max_turns=int(ev.get("max_turns", 240)), limits=limits, generation_constraints=gen, adapter_manifest=adapters,
                              context_cache_config=ContextCacheConfig(min_tokens=2048, ttl_seconds=1800, cache_intervals=10),
                              events_compaction_config=EventsCompactionConfig(compaction_interval=15, overlap_size=2, token_threshold=14336, event_retention_size=5),
                              graph_dir=str(OUT / "nograph"), embeddings_dir=str(OUT / "nograph"), wheels_dir=None, verbose=False, display_mode="quiet")
            t1 = time.time()
            r = run_sync(lambda: Evaluator(ecfg).evaluate_task(task=task, task_index=1, total_tasks=1))
            patch = (r.agent_patch or "")
            (res_dir / "agent.patch").write_text(patch)
            row = dict(model=label, task=task.instance_id, patch_chars=len(patch), tool_calls=r.tool_calls, minutes=round((time.time() - t1) / 60, 1),
                       error=getattr(r, "agent_error", None))
            results.append(row)
            print(row)
            if after_each:
                after_each(task, res_dir)
    finally:
        server.stop()
    return results


def apply_patch(patch_file, code_dir):
    code_dir.mkdir(parents=True, exist_ok=True)
    if not (code_dir / ".git").exists():
        subprocess.run(["git", "init", "-q", "-b", "main"], cwd=code_dir, check=True)
    if Path(patch_file).read_text().strip():
        p = subprocess.run(["git", "apply", "--whitespace=nowarn", "--reject", str(patch_file)], cwd=code_dir, capture_output=True, text=True)
        print("patch ->", code_dir, "ok" if p.returncode == 0 else p.stderr[:500])


def audit(label, task_id, forbidden):
    hits = []
    for f in (OUT / "results" / label / task_id).rglob("*.log"):
        txt = f.read_text(errors="ignore")
        hits += [s for s in forbidden if s and s in txt]
    return sorted(set(hits))


def write_tasks_file(tasks):
    with open(OUT / "tasks.jsonl", "w") as f:
        for t in tasks:
            f.write(json.dumps(dict(instance_id=t.instance_id, repo=t.repo, base_commit=t.base_commit, problem_statement=t.problem_statement)) + "\n")
```

### Custom Gymnasium Environment & RL Agent for DICOM-USD Alignment

Below, we define a complete RL structure using `Stable-Baselines3` principles (or a pure PyTorch optimization loop if dependencies are tight) to map raw DICOM/NII slices to the oriented, healthy USD archetype. The transformation agent outputs a state-representation Z-vector mapped between `[-1, 1]`.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.distributions import Normal

class SoftmaxZGenerator(nn.Module):
    """
    Generates a Z vector mapped from -1 to 1 via a softmax-weighted projection.
    """
    def __init__(self, input_dim=512, z_dim=12):
        super().__init__()
        self.fc = nn.Linear(input_dim, z_dim)
        # Fixed scale grid matching the -1 to 1 distribution boundary
        self.register_buffer("grid", torch.linspace(-1.0, 1.0, z_dim))

    def forward(self, x):
        logits = self.fc(x)
        weights = F.softmax(logits, dim=-1)
        # Weighted sum of the scale grid to project strictly into [-1, 1]
        z_vector = torch.sum(weights * self.grid, dim=-1, keepdim=True)
        return z_vector, weights

class KneeAlignmentPolicy(nn.Module):
    """
    Deep Actor-Critic Network guiding the physical alignment (rotation, translation)
    of the test NII/DICOM relative to the healthy USD archetype.
    """
    def __init__(self, z_dim=12):
        super().__init__()
        # Backbone to extract features from input slice image
        self.feature_extractor = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((8, 8)),
            nn.Flatten()
        )
        self.z_generator = SoftmaxZGenerator(input_dim=64*8*8, z_dim=z_dim)
        self.actor = nn.Linear(z_dim, 6) # Outputs actions: 3D rotation and translation scale
        self.critic = nn.Linear(z_dim, 1)

    def forward(self, state_slice):
        features = self.feature_extractor(state_slice)
        z_vector, weights = self.z_generator(features)

        action_mean = self.actor(weights) # Control shifts
        value = self.critic(weights)
        return action_mean, z_vector, value
```

### Divergence & Voxel Weighting Mechanism
This cell computes the voxel-wise anomaly map by highlighting regions that diverge significantly from the baseline archetype.

```python
def compute_voxel_divergence(test_volume, archetype_volume, z_weights):
    """
    Computes a voxel weighted divergence map scaled to [-1, 1].
    -1 represents highly abnormal/divergent voxels relative to the healthy USD baseline.
    """
    # Ensure tensor computation
    t_vol = torch.tensor(test_volume, dtype=torch.float32)
    a_vol = torch.tensor(archetype_volume, dtype=torch.float32)

    # Raw voxel-level structural absolute difference
    raw_diff = torch.abs(t_vol - a_vol)

    # Generate probability map using softmax over spatial dimensions
    flat_diff = raw_diff.view(-1)
    spatial_softmax = F.softmax(flat_diff, dim=0).view_as(raw_diff)

    # Scale to [-1, 1] range: high divergence maps to -1
    divergence_map = 1.0 - 2.0 * (spatial_softmax / (spatial_softmax.max() + 1e-8))
    return divergence_map.numpy()
```
